### Evaluate baseline

````python
from ragas import metrics, RunConfig
from ragas.metrics.base import Metric


llm_model: str = "gpt-4o-mini"
embeddings_model: str = "text-embedding-ada-002"
question_type: str = 'novel-generated'
our_metrics: list[Metric] = [
    metrics.answer_correctness,
    metrics.faithfulness,
    metrics.context_recall,
    metrics.context_precision
]

run_config = RunConfig(
    timeout=500, 
    log_tenacity=True,
    max_wait=120,
    max_workers=4,
) 
````

````python
from datasets import load_dataset

dataset = load_dataset(
    "PatronusAI/financebench", 
    split="train",
    cache_dir="dataset" 
)

dataset
````

````python
# Фільтруємо тільки novel-generated (Унікальні / Синтетичні)

# специфічні питання, згенеровані (зазвичай LLM) спеціально під унікальний контекст конкретного документа.
# Що тестує у RAG:
#  - Reasoning (Логіку): Чи може RAG знайти не просто ключові слова, а пояснення події.
#  - Стійкість до рідкісних запитів: Це перевірка того, наскільки добре система справляється з нестандартними запитами.

novel_subset = dataset.filter(lambda x: x['question_type'] == question_type)

print(
    f"- Загальний розмір датасету: {len(dataset)}.\n"
    f"- Обсяг тестової вибірки: {len(novel_subset)} питань (підмножина {question_type}).\n"
    f"- Кількість документів у БД: {len(sorted(set(novel_subset['doc_link'])))} унікальних PDF-документів.\n"
    f"- Умови оцінювання (Суддя): Ragas + LLM ({llm_model}) + OpenAIEmbeddings ({embeddings_model})\n"
    f"- Збалансованість та складність: 'novel-generated' - це синтетичні питання під унікальний контекст.\n"
    "\tВони тестують глибоку логіку (Reasoning) та стійкість до нестандартних запитів.\n"
    "\tЦе найскладніший тип питань для RAG."
)
````

````python
import os
import requests

from dotenv import load_dotenv

from chatbot.models.chat import Chat
from chatbot.globals import DEFAULT_API_HOST, api_settings

load_dotenv()

os.environ["RAGAS_DEBUG"] = "true"
API_URL = f"http://{DEFAULT_API_HOST}:{api_settings.PORT}"

results_buffer = {
    "question": [],
    "answer": [],      # Відповідь RAG
    "contexts": [],    # Чанки, які знайшов RAG
    "ground_truth": [] # Еталон з датасету
}
````

````python
for row in novel_subset:
    question = row['question']
    ground_truth = row['answer']
    
    response = requests.get(
        f"{API_URL}/chat/query_baseline",
        params={
            "user_input": question,
        },
        timeout=None,
    )

    response.raise_for_status()

    data: Chat = Chat(**response.json())
    print(data.llm_response)
    
    results_buffer["question"].append(question)
    results_buffer["answer"].append(data.llm_response)
    results_buffer["contexts"].append([chunk.chunk.content for chunk in data.rag_results.retrieved_chunks])
    results_buffer["ground_truth"].append(ground_truth)
````

````python
import pandas as pd

results_buffer_df = pd.DataFrame(results_buffer)
results_buffer_df.to_csv("data/baseline_novel_generated_results_buffer.csv", index=False)
results_buffer_df
````

````python
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from ragas import evaluate
from datasets import Dataset


eval_results = evaluate(
    dataset=Dataset.from_dict(results_buffer),
    metrics=our_metrics,
    llm=ChatOpenAI(model=llm_model),
    embeddings=OpenAIEmbeddings(model=embeddings_model),
    run_config=run_config
)

print(eval_results)
df = eval_results.to_pandas()
df.to_csv("data/baseline_novel_generated_results.csv")
````

### Results

#### Explain metrics:
- **Context Recall** (повнота пошуку)
    - *Визначення:* Міра здатності системи знайти всі необхідні фрагменти інформації, які потрібні для відповіді на питання.
    - *Як це працює:* Порівнює знайдені вашою системою чанки (contexts) з еталонною відповіддю (ground_truth). Якщо в еталоні згадується "прибуток за 2021 і 2022 роки", а ваш ретрівер знайшов дані тільки за 2021, Recall буде низьким (близько 0.5).
- **Context Precision** (Точність контексту)
    - *Визначення:* Відношення кількості релевантних (корисних) чанків до загальної кількості чанків, переданих у LLM. Також враховує ранжування (чи знаходяться корисні чанки нагорі списку).
    - *Як це працює:* Якщо ви передали в LLM 10 фрагментів тексту, але тільки 2 з них містять відповідь, а 8 — це "сміття" (вода, колонтитули, інша тема), то Precision буде низьким (0.2).
- **Faithfulness** (Вірність / Фактологічна стійкість)
    - *Визначення:* Міра того, наскільки згенерована відповідь базується виключно на наданому контексті.
    - *Як це працює:* [RAGAS](https://docs.ragas.io/en/stable/) розбиває відповідь LLM на окремі твердження і перевіряє, чи підтверджується кожне з них знайденими чанками (contexts). Якщо модель відповідає правильно, але використовує знання, яких не було в контексті (свої власні "тренувальні" знання), або просто вигадує факти — Faithfulness падає.
- **Answer Correctness** (Коректність відповіді)
    - *Визначення:* Комплексна оцінка, яка вимірює семантичну та фактологічну схожість згенерованої відповіді (answer) до еталонної відповіді експерта (ground_truth).
    - *Як це працює:* Враховує два аспекти:
        - *Семантична схожість:* Чи однаковий сенс у текстів?
        - *Фактологічна точність:* Чи збігаються конкретні цифри, дати та сутності?
- Більше інформації в [документації](https://docs.ragas.io/en/stable/concepts/metrics/overview/#2-component-level-metrics)

## [Hotpot QA](https://huggingface.co/datasets/hotpotqa/hotpot_qa)

````python
from datasets import load_dataset
hotpot = load_dataset(
    "hotpot_qa", 
    "distractor",
    cache_dir="dataset" 
)
````

````python
train = hotpot["train"]
train
````

````python
train[0]["context"].items()
````

````python
num_samples = 20  
contexts_for_db = []
test_cases = []

for idx, example in enumerate(hotpot["train"]):
    if idx >= num_samples:
        break
    
    question = example["question"]
    answer = example["answer"]
    supporting_facts = example["supporting_facts"]
    
    titles = example["context"]["title"]
    sentences_list = example["context"]["sentences"]
    
    unique_contexts = []
    
    for title, sentences in zip(titles, sentences_list):
        if not sentences:
            continue
        
        # ОБ'ЄДНУЙте ВСІ речення з цього документу в ОДИН контекст
        full_context = " ".join(sentences)
        
        if full_context.strip():
            is_supporting = any(title == sf[0] for sf in supporting_facts)
            
            unique_contexts.append({
                "content": full_context,
                "source_title": title,
                "is_supporting": is_supporting,
                "example_id": idx,
                "question_id": idx
            })
    
    contexts_for_db.extend(unique_contexts)
    
    test_cases.append({
        "question_id": idx,
        "question": question,
        "answer": answer,
        "contexts": unique_contexts,
        "supporting_facts": supporting_facts
    })

print(f"Контекстів: {len(contexts_for_db)}")
print(f"Тестових випадків: {len(test_cases)}")
````
